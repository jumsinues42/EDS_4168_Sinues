# STEP 4: DATA CLEANING

```python
# Why? Real-world data is messy! We need to clean it first

print("\n🧹 DATA CLEANING STARTED...")
# Remove extra spaces from column names
# Why? Spaces can cause errors when accessing columns
df.columns = df.columns.str.strip()

# Check for missing values
# Why? Missing data can break our analysis
print("\n❓ Missing Values Check:")
print(df.isnull().sum())
```

```python
# Clean 'Intelligence Index' column (has 'E' and '--' values)
# Why? Need numeric values for calculations
def clean_intelligence(value):
    """Convert intelligence scores to numbers"""
    if isinstance(value, str):
        value = value.strip()
        if value == '--' or value == '':
            return np.nan  # Replace '--' with NaN (Not a Number)
        value = value.replace('E', '').strip()  # Remove 'E' marker
    try:
        return float(value)
    except:
        return np.nan

df['Intelligence Index'] = df['Intelligence Index'].apply(clean_intelligence)
```

```python
df['Intelligence Index']
```

```python
# Clean 'Price' column (remove $ and convert to float)
# Why? Can't do math on text like "$4.81"
df['Price (Blended USD/1M Tokens)'] = df['Price (Blended USD/1M Tokens)'].str.replace('$', '').str.strip()
df['Price (Blended USD/1M Tokens)'] = pd.to_numeric(df['Price (Blended USD/1M Tokens)'], errors='coerce')
```

```python
df['Price (Blended USD/1M Tokens)']
```

```python
# Rename columns for easier access
# Why? Shorter names = less typing!
df.rename(columns={
    'Price (Blended USD/1M Tokens)': 'Price_USD',
    'Speed(median token/s)': 'Speed_TPS',
    'Latency (First Answer Chunk /s)': 'Latency_Sec',
    'Context Window': 'Context_Window',
    'Intelligence Index': 'Intelligence'
}, inplace=True)

print("✅ Data cleaning completed!")
print("="*60)
```

# QUESTION 2: What's the intelligence distribution?

```python
# QUESTION 2: What's the intelligence distribution?
# ============================================================
print("\n🧠 QUESTION 2: How intelligent are these AI models?")

# Remove NaN values for this analysis
intel_clean = df['Intelligence'].dropna()

print(f"Average Intelligence Score: {intel_clean.mean():.2f}")
print(f"Median Intelligence Score: {intel_clean.median():.2f}")
print(f"Highest Score: {intel_clean.max():.2f}")
print(f"Lowest Score: {intel_clean.min():.2f}")
```

```python

plt.figure(figsize=(12, 5))

# Histogram: Shows distribution (how many models in each range)
plt.subplot(1, 2, 1)
plt.hist(intel_clean, bins=20, color='purple', alpha=0.7, edgecolor='black')
plt.title('Intelligence Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Intelligence Score', fontsize=10)
plt.ylabel('Number of Models', fontsize=10)
plt.axvline(intel_clean.mean(), color='red', linestyle='--', label=f'Mean: {intel_clean.mean():.1f}')
plt.legend()
```

**Intelligence Distribution Insights:**

- Most AI models cluster around an intelligence score of 15-20, showing that the majority fall into a mid-range capability level. 
- The mean intelligence score is 21.7, indicated by the red dashed line. 
- Very few models achieve high intelligence scores above 40, suggesting that creating highly intelligent AI systems is still quite rare and challenging. 
- The distribution is right-skewed, meaning there are more models with lower to moderate intelligence than highly advanced ones.

```python
# Box Plot: Shows outliers and quartiles
plt.subplot(1, 2, 2)
plt.boxplot(intel_clean, vert=True)
plt.title('Intelligence Box Plot', fontsize=12, fontweight='bold')
plt.ylabel('Intelligence Score', fontsize=10)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

**Intelligence Box Plot Insights:**

- The median intelligence score (the line inside the box) is around 20, meaning half of all AI models score below this and half above. 
- The middle 50% of models have intelligence scores between roughly 14 and 29 (the box range), showing moderate variation in capabilities. 
- There's one outlier at approximately 52, representing an exceptionally intelligent AI model that far exceeds typical performance. 
- The whiskers extend from about 6 to 49, indicating that while most models cluster in the 14-29 range, there's significant diversity in AI intelligence across the full dataset.

# QUESTION 3: Price Analysis

```python
# QUESTION 3: Price Analysis
# ============================================================
print("\n💰 QUESTION 3: How much do AI models cost?")

price_clean = df['Price_USD'].dropna()

print(f"Average Price: ${price_clean.mean():.2f} per 1M tokens")
print(f"Median Price: ${price_clean.median():.2f} per 1M tokens")
print(f"Most Expensive: ${price_clean.max():.2f}")
print(f"Cheapest: ${price_clean.min():.2f}")
```

```python
# Find models in different price ranges
budget = len(price_clean[price_clean < 0.50])
mid = len(price_clean[(price_clean >= 0.50) & (price_clean < 2.00)])
premium = len(price_clean[price_clean >= 2.00])

print(f"\n💵 Budget Models (<$0.50): {budget} models")
print(f"💵 Mid-Range ($0.50-$2.00): {mid} models")
print(f"💵 Premium (>$2.00): {premium} models")
```

```python
# Histogram with log scale (because prices vary a LOT)
plt.subplot(1, 2, 1)
plt.hist(price_clean, bins=30, color='green', alpha=0.7, edgecolor='black')
plt.title('Price Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Price (USD per 1M tokens)', fontsize=10)
plt.ylabel('Number of Models', fontsize=10)
plt.yscale('log')  # Log scale to see small values better
```

**Price Distribution Insights:**

- **Most AI models are very affordable**: The vast majority (over 100 models) are priced under $1 per 1 million tokens, making AI accessible to most users and businesses.

- **Logarithmic scale reveals pricing tiers**: The y-axis uses a log scale, showing that while hundreds of models cluster at low prices, there are distinct pricing tiers as capabilities increase.

- **Premium models exist but are rare**: Only a handful of models cost between $4-10 per 1M tokens, suggesting these are likely the most advanced or specialized AI systems with premium pricing.

- **Sharp drop-off after $1**: There's a dramatic decrease in the number of models above $1, indicating that the AI market is highly competitive in the budget-to-moderate price range, with fewer options at premium price points.

# QUESTION 5: Intelligence vs Price (Trade-off)

```python
print("\n🎯 QUESTION 5: Does higher intelligence mean higher price?")

# Create clean dataset with both values
trade_off = df[['Intelligence', 'Price_USD', 'Model', 'Creator']].dropna()

# Calculate correlation
# Correlation: -1 to 1 (1 = perfect positive, -1 = perfect negative, 0 = no relation)
correlation = trade_off['Intelligence'].corr(trade_off['Price_USD'])
print(f"Correlation between Intelligence & Price: {correlation:.3f}")

if abs(correlation) < 0.3:
    print("📊 Weak correlation - Price doesn't strongly depend on intelligence")
elif abs(correlation) < 0.7:
    print("📊 Moderate correlation - Some relationship exists")
else:
    print("📊 Strong correlation - Price and intelligence are related")
```

```python
# Scatter plot: Each dot is one model
plt.scatter(trade_off['Intelligence'], trade_off['Price_USD'], 
           alpha=0.6, s=100, c=trade_off['Intelligence'], 
           cmap='viridis', edgecolors='black')

plt.colorbar(label='Intelligence Score')
plt.title('Intelligence vs Price Trade-off', fontsize=14, fontweight='bold')
plt.xlabel('Intelligence Score', fontsize=12)
plt.ylabel('Price (USD per 1M tokens)', fontsize=12)
plt.grid(True, alpha=0.3)

# Add trend line
z = np.polyfit(trade_off['Intelligence'], trade_off['Price_USD'], 1)
p = np.poly1d(z)
plt.plot(trade_off['Intelligence'], p(trade_off['Intelligence']), 
         "r--", alpha=0.8, label='Trend Line')
plt.legend()

plt.tight_layout()
plt.show()
```

**Intelligence vs Price Trade-off Insights:**

- **Positive correlation exists**: The red trend line shows that higher intelligence scores generally correlate with higher prices, confirming users pay premium for smarter AI models.

- **Budget options dominate low intelligence**: The dense cluster of dark blue/purple dots at intelligence scores below 20 and prices under $1 shows abundant affordable options for basic AI needs.

- **Diminishing returns above mid-range**: Models with intelligence scores of 30-40 show wide price variation ($1-$6), suggesting that incremental intelligence gains become increasingly expensive and less predictable.

- **Elite models command premium**: The yellow-green dots at intelligence 40-50 and prices $3-$10 represent cutting-edge AI, where top performance justifies significantly higher costs for specialized applications.

- **Trade-off exists in lower-tier models**: Orange dots show that many models sacrifice one metric for the other—some achieve moderate speed with high latency (up to 110s), while others maintain low latency at slower speeds.

- **Most models cluster at low latency**: The dense concentration of both orange and green markers below 20 seconds latency indicates that modern AI prioritizes responsiveness, with few models tolerating delays above 30s.

- **Speed doesn't guarantee low latency**: Models at 50-100 t/s show latency ranging from 1s to 45s, proving that raw throughput speed and response time are distinct optimization challenges requiring different technical approaches.
